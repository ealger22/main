```r
# ---------------------------------------------------
# 📊 R Script: Analysis of mtcars Dataset
# ---------------------------------------------------
# Description: This script performs statistical analysis on the mtcars dataset,
# including mean and standard deviation calculations, column summaries,
# and specific data retrieval for Mazda RX4.
# ---------------------------------------------------

# Load the built-in mtcars dataset
data(mtcars)

# ------------------ Step 1: Column & Row Names ------------------
cat("\n🔹 Column Names in mtcars Dataset:\n")
print(colnames(mtcars))

cat("\n🔹 Row Names in mtcars Dataset:\n")
print(rownames(mtcars))

# ------------------ Step 2: Mean & Standard Deviation ------------------
# Calculate mean & standard deviation for mpg and cyl
mpg_mean <- mean(mtcars$mpg)
mpg_sd <- sd(mtcars$mpg)

cyl_mean <- mean(mtcars$cyl)
cyl_sd <- sd(mtcars$cyl)

# Display results
cat("\n📊 Statistical Summary for mpg & cyl:\n")
cat(sprintf("✔️ Mean of mpg: %.2f\n✔️ Standard Deviation of mpg: %.2f\n", mpg_mean, mpg_sd))
cat(sprintf("✔️ Mean of cyl: %.2f\n✔️ Standard Deviation of cyl: %.2f\n", cyl_mean, cyl_sd))

# ------------------ Step 3: Column Means for All Variables ------------------
cat("\n📈 Average Values for All Numeric Variables:\n")
print(colMeans(mtcars))

# ------------------ Step 4: Retrieve cyl Value for Mazda RX4 ------------------
mazda_cyl <- mtcars["Mazda RX4", "cyl"]
cat("\n🚗 Mazda RX4 - Cylinders Value:\n")
cat(sprintf("✔️ Mazda RX4 cyl: %d\n", mazda_cyl))

# ------------------ End of Script ------------------
cat("\n✅ Analysis Complete. Script executed successfully!\n")
```
