```python
# ✅ Step 1: Mount Google Drive
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')

# ✅ Step 2: Define File Path (Modify if needed)
file_path = "/content/drive/My Drive/lego_setsHB.csv"

# ✅ Step 3: Read Dataset
df = pd.read_csv(file_path)

# ✅ Step 4: Dataset Overview
print("\n🔍 **Dataset Overview**")
print("--------------------------------------------------")
df.info()

print("\n📊 **Dataset Summary Statistics**")
print("--------------------------------------------------")
print(df.describe())

# ✅ Step 5: Filter LEGO Sets with Star Rating >= 4
high_rating_lego_sets = df[df["star_rating"] >= 4]

# ✅ Step 6: Count the Number of High-Rated LEGO Sets
high_rating_count = high_rating_lego_sets["star_rating"].count()

# ✅ Step 7: Calculate the Average List Price for High-Rated Sets
average_price_high_rating = high_rating_lego_sets["list_price"].mean()

# ✅ Step 8: Display Clean Output
print("\n🎯 **LEGO Data Analysis Results**")
print("--------------------------------------------------")
print(f"✔️ Total LEGO Sets: {len(df):,}")
print(f"✔️ High-Rated LEGO Sets (⭐ ≥ 4.0): {high_rating_count:,}")
print(f"✔️ Average Price of High-Rated LEGO Sets: ${average_price_high_rating:.2f}")

# ✅ Step 9: Show a Sample of High-Rated LEGO Sets
print("\n📌 **Sample LEGO Sets with ⭐ Rating ≥ 4.0**")
print("--------------------------------------------------")
display(high_rating_lego_sets.head(10))  # Displays first 10 rows in Colab-friendly format
```
