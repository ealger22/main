```python
# ---------------------------------------------------
# 📊 Titanic Dataset Analysis (R Script)
# ---------------------------------------------------
# Description:
# This script analyzes the Titanic dataset, performing:
# - Statistical summary analysis
# - Data cleaning (removing NA values)
# - Filtering passengers based on age and survival status
# ---------------------------------------------------

# Load necessary libraries
library(dplyr)

# ------------------ Step 1: Load Dataset ------------------
# Define dataset path (Ensure the file is in Google Drive)
dataset_path <- "/content/drive/MyDrive/Titanic-Dataset.csv"

# Read Titanic dataset into a DataFrame
df <- read.csv(dataset_path, stringsAsFactors = TRUE)

# Confirm dataset loaded successfully
cat("\n✅ Titanic dataset loaded successfully. Total rows:", nrow(df), "\n")

# ------------------ Step 2: Statistical Summary ------------------
cat("\n📊 Generating statistical summary of the dataset:\n")
summary(df)

# Highlight key numerical columns with meaningful statistics:
# - "Age"  (Mean, Median)
# - "Fare" (Mean, Median)
# - "SibSp" (Siblings/spouses aboard)
# - "Parch" (Parents/children aboard)
# - "Pclass" (Passenger class distribution)

# ------------------ Step 3: Remove Rows with Missing Age ------------------
df2 <- df[!is.na(df$Age), ]  # Remove NA values in Age column
cat("\n✅ Rows without missing Age assigned to df2. Total rows:", nrow(df2), "\n")

# ------------------ Step 4: Filter Survived Passengers by Age ------------------
# Passengers under 30 who survived
younger <- df2[df2$Age < 30 & df2$Survived == 1, ]
cat("\n🚸 Passengers Under 30 Who Survived: ", nrow(younger), "\n")

# Passengers 30 and older who survived
older <- df2[df2$Age >= 30 & df2$Survived == 1, ]
cat("\n🧑‍🦳 Passengers 30+ Who Survived: ", nrow(older), "\n")

# ------------------ End of Analysis ------------------
cat("\n✅ Analysis Completed Successfully! 🎯\n")
```

```python
%%R
# Load necessary libraries
library(dplyr)

# Define dataset path
dataset_path <- "/content/drive/MyDrive/Titanic-Dataset.csv"

# Read Titanic dataset
df <- read.csv(dataset_path, stringsAsFactors = TRUE)

# ------------------ Step 2: Statistical Summary ------------------
cat("\n📊 Statistical Summary of Titanic Dataset:\n")
summary(df)

# Columns with meaningful statistical summaries:
# - "Age" (Mean, Median)
# - "Fare" (Mean, Median)
# - "SibSp" (Number of siblings/spouses aboard)
# - "Parch" (Number of parents/children aboard)
# - "Pclass" (Passenger class distribution)

# ------------------ Step 3: Remove NA values from Age column ------------------
df2 <- df[!is.na(df$Age), ]

cat("\n✅ Rows with non-NA Age values assigned to df2. Total rows in df2:", nrow(df2), "\n")

# ------------------ Step 4: Assign Survived Passengers Under 30 to younger ------------------
younger <- df2[df2$Age < 30 & df2$Survived == 1, ]

cat("\n🚸 Passengers Under 30 Who Survived Assigned to younger. Total:", nrow(younger), "\n")

# ------------------ Step 5: Assign Survived Passengers Aged 30+ to older ------------------
older <- df2[df2$Age >= 30 & df2$Survived == 1, ]

cat("\n🧑‍🦳 Passengers Aged 30+ Who Survived Assigned to older. Total:", nrow(older), "\n")

# ------------------ End of Script ------------------
cat("\n✅ Analysis Complete. Script executed successfully!\n")
```
